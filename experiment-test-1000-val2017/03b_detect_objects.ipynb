# 03b — (Optional) DETR detections for query metadata without an oracle

```text
(OPTIONAL) DETR object detections -> 'realistic' query-side metadata.
BY DEFAULT: val + test only (200 images, a few minutes on CPU). The KB retains COCO annotations;
to run detection on the KB as well (10,000 images, several hours on CPU; use a GPU/Colab): --splits kb val test.

Why: in 03, objects come from COCO annotations (ground truth). For a test image, using the
ground truth as a filtering query is an oracle. Metadata-Filtered RAG groups (G28-G30) must use THESE
detections for val/test queries (and may use GT or detections for the KB; document the choice).

Output: tables/detections_detr.jsonl  {image_id, split, detections:[{category, score, bbox_norm}]}, threshold = config.detector.conf_threshold
Note: requires torch + transformers + access to huggingface.co (not tested offline in the development environment).
```

> Project: a comparative study of RAG techniques for image captioning (MS COCO) — **Group G1**. Run the notebooks in order: 00 → 01 → 01b → 02 → … → 09. Shared helpers are in `common.py`, `visual.py`, and `rag_data.py`.

````python
import os
# --- Configuration: use "config.yaml" for real COCO data; use "config_test.yaml" for the synthetic mini-COCO dataset (offline testing).
CONFIG = os.environ.get("RAG_CONFIG", "config.yaml")
os.environ["RAG_CONFIG"] = CONFIG
print("Configuration used:", CONFIG)
````

## Imports et constantes

````python
import argparse
import numpy as np
import pandas as pd
from PIL import Image
from common import get_logger, load_config, mkdirs, paths, write_json, write_jsonl
log = get_logger("detr")
````

### `main()`

````python
def main():
    ap = argparse.ArgumentParser(); ap.add_argument("--batch", type=int, default=16)
    ap.add_argument("--splits", nargs="*", default=["val", "test"], choices=["kb", "val", "test"])
    args = ap.parse_args()
    import torch
    from transformers import AutoImageProcessor, DetrForObjectDetection
    cfg = load_config(); p = paths(cfg); mkdirs(p)
    dev = "cuda" if torch.cuda.is_available() else "cpu"
    proc = AutoImageProcessor.from_pretrained(cfg["detector"]["model"])
    model = DetrForObjectDetection.from_pretrained(cfg["detector"]["model"]).to(dev).eval()
    thr = cfg["detector"]["conf_threshold"]
    imgs = pd.read_csv(p["tables"] / "images.csv"); imgs = imgs[(imgs.status == "ok") & imgs.split.isin(args.splits)].sort_values(["split", "image_id"])
    log.info("Detected splits: %s -> %d images", args.splits, len(imgs))
    out = []
    rows = list(imgs.itertuples())
    for i in range(0, len(rows), args.batch):
        b = rows[i:i + args.batch]
        ims = [Image.open(p["art"] / r.path_rgb).convert("RGB") for r in b]
        with torch.no_grad():
            inp = proc(images=ims, return_tensors="pt").to(dev)
            o = model(**inp)
        res = proc.post_process_object_detection(o, threshold=thr, target_sizes=torch.tensor([im.size[::-1] for im in ims]).to(dev))
        for r, im, d in zip(b, ims, res):
            W, H = im.size
            dets = [{"category": model.config.id2label[int(l)], "score": round(float(s), 4),
                     "bbox_norm": [round(float(v), 4) for v in (bx[0] / W, bx[1] / H, bx[2] / W, bx[3] / H)]}
                    for s, l, bx in zip(d["scores"].cpu(), d["labels"].cpu(), d["boxes"].cpu())]
            out.append({"image_id": int(r.image_id), "split": r.split, "detections": dets})
        if (i // args.batch) % 5 == 0:
            log.info("%d/%d", i, len(rows))
    write_jsonl(p["tables"] / "detections_detr.jsonl", out)
    write_json(p["audit"] / "detector_info.json", {"model": cfg["detector"]["model"], "conf_threshold": thr, "n_images": len(out), "splits": args.splits,
                                                  "note": "DETR COCO: its 91 IDs/80 classes match COCO classes and are comparable to the GT annotations."})
````

## Run this step

````python
import sys
sys.argv = ["notebook", "--splits", "val", "test"]          # No CLI arguments are used in a notebook.
try:
    main()
except SystemExit as e:               # Some scripts exit through sys.exit(code).
    print("Finished with exit code:", e.code)
````
